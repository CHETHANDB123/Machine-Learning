##Hands on Linear SVM

```python
import pandas as pd
data = pd.read_csv("/content/apples_and_oranges.csv")
data.head()
```

```python
data.shape
```

```python
data.info()
```

```python
data.describe()
```

```python
data.isnull().sum()
```

###Treating Categorial Variable

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
data.Class = le.fit_transform(data.Class)
data.head()
```

###Sampling the Dataset

```python
from sklearn.model_selection import train_test_split
x=data[['Weight','Size']]
y=data.Class
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=2)
```

###Building the Linear SVC Classifier

```python
import warnings
warnings.simplefilter("ignore")
from sklearn.svm import LinearSVC
classifier = LinearSVC()
classifier.fit(x_train,y_train)
```

###Calculating Test accuracy

```python
from sklearn.metrics import accuracy_score
y_pred = classifier.predict(x_test)
print("Test Accuracy=",accuracy_score(y_test,y_pred))
```

###Confusion Matrix

```python
from sklearn.metrics import classification_report,confusion_matrix
print(confusion_matrix(y_test,y_pred))
```

##Hands on for NON Linear SVM

```python
import pandas as pd
df = pd.read_csv("/content/Iris.csv")
df.head()
```

```python
df.shape
```

```python
df.describe()
```

```python
df.isnull().sum()
```

###Build Multiclass Linear & Non-Linear Models

```python
from sklearn.svm import LinearSVC, SVC
x = df.iloc[:,0:5]
y = df.Species

lin_svc = LinearSVC().fit(x,y)

# svc with linear kernel
svc_lin = SVC(kernel='linear').fit(x,y)
# the linear kernel assumes linear decision boundary

# svc with RBF kernel
rbf_svc = SVC(kernel='rbf').fit(x,y)

poly_svc3 = SVC(kernel='poly',degree=3).fit(x,y)

poly_svc4 = SVC(kernel='poly',degree=4).fit(x,y)
```

###Printing all the model scores

```python
print("SVC Linear Score:",svc_lin.score(x,y))
print("RBF SVC Score:",rbf_svc.score(x,y))
print("3 Degree Polynomial SVC Score:",poly_svc3.score(x,y))
print("4 Degree Polynomial SVC Score:",poly_svc4.score(x,y))
```
