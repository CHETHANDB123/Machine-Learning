#Measures of Spread
##Measures of spread refer to how spread out or dispersed the values in a dataset are. Common measures of spread include the range, variance, standard deviation, and interquartile range (IQR).


Range: The range is the difference between the maximum and minimum values. It gives a simple measure of the spread, but it's sensitive to outliers.

Variance: Variance measures the average of the squared differences from the mean. It's useful for understanding how data points deviate from the mean.

Standard Deviation: The square root of variance, it represents spread in the same units as the data, making it more interpretable.


Interquartile Range (IQR): The IQR measures the middle 50% of the data and is less sensitive to outliers compared to range or variance.

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Generating a sample data
np.random.seed(42)
data = np.random.normal(loc=50, scale=15, size=1000)

# Calculating measures of spread
range_value = np.ptp(data)
variance = np.var(data)
std_dev = np.std(data)
iqr_value = np.percentile(data, 75) - np.percentile(data, 25)

# Plotting the data distribution
plt.figure(figsize=(8, 6))
sns.histplot(data, kde=True, color='skyblue', bins=30)
plt.title("Distribution of Data")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()

# Displaying the calculated measures
print(f"Range: {range_value}")
print(f"Variance: {variance}")
print(f"Standard Deviation: {std_dev}")
print(f"Interquartile Range (IQR): {iqr_value}")
```

#2. Measure of Centroid
The centroid (or center) of a dataset typically refers to its mean, median, or mode, which are central tendencies.

Mean: The arithmetic average of the dataset, useful in symmetric distributions but sensitive to outliers.

Median: The middle value when data is sorted, useful for skewed distributions as it's robust to outliers.

Mode: The most frequent value, useful for categorical data and skewed distributions.

```python

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

# Generate a sample data
np.random.seed(42)
data = np.random.normal(loc=50, scale=15, size=1000)

# Calculate measures of spread
range_value = np.ptp(data)
variance = np.var(data)
std_dev = np.std(data)
iqr_value = np.percentile(data, 75) - np.percentile(data, 25)

# # Plotting the data distribution
# plt.figure(figsize=(8, 6))
# sns.histplot(data, kde=True, color='skyblue', bins=30)
# plt.title("Distribution of Data")
# plt.xlabel("Values")
# plt.ylabel("Frequency")
# plt.show()

# # Displaying the calculated measures
# print(f"Range: {range_value}")
# print(f"Variance: {variance}")
# print(f"Standard Deviation: {std_dev}")
# print(f"Interquartile Range (IQR): {iqr_value}")


# Calculate mean, median, and mode
mean_value = np.mean(data)
median_value = np.median(data)
#Using scipy stats for mode calculation to handle multimodal distributions
mode_value = stats.mode(data)

# Plotting the data with centroid markers
plt.figure(figsize=(8, 6))
sns.histplot(data, kde=True, color='lightgreen', bins=30)
plt.axvline(mean_value, color='red', linestyle='--', label=f'Mean: {mean_value:.2f}')
plt.axvline(median_value, color='blue', linestyle='-', label=f'Median: {median_value}')


plt.title("Data with Mean, Median, and Mode")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.legend()
plt.show()

# Displaying the centroid measures
print(f"Mean: {mean_value}")
print(f"Median: {median_value}")
```

#  4 Major Distributions
The four major distributions are Normal, Uniform, Exponential, and Binomial. Each has different characteristics and use cases.

Normal Distribution: Bell-shaped and symmetric around the mean, with the empirical rule (68-95-99.7 rule) for standard deviations.

Uniform Distribution: Every outcome has an equal probability of occurring within a defined range.


Exponential Distribution: Often models waiting times or the time between events in a Poisson process, with a skewed shape.

`
Binomial Distribution: Models the number of successes in a fixed number of independent trials, with only two possible outcomes.

```python
import scipy.stats as stats

# Generating data for the distributions
normal_data = np.random.normal(loc=0, scale=1, size=1000)
uniform_data = np.random.uniform(low=0, high=10, size=1000)
exponential_data = np.random.exponential(scale=1, size=1000)
binomial_data = np.random.binomial(n=10, p=0.5, size=1000)

# Plotting the distributions
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
sns.histplot(normal_data, kde=True, color='blue', bins=30)
plt.title("Normal Distribution")

plt.subplot(2, 2, 2)
sns.histplot(uniform_data, kde=True, color='green', bins=30)
plt.title("Uniform Distribution")

plt.subplot(2, 2, 3)
sns.histplot(exponential_data, kde=True, color='red', bins=30)
plt.title("Exponential Distribution")

plt.subplot(2, 2, 4)
sns.histplot(binomial_data, kde=True, color='purple', bins=30)
plt.title("Binomial Distribution")

plt.tight_layout()
plt.show()
```

#5. Probability
Probability is a measure of the likelihood of an event occurring. It ranges from 0 to 1.

```python
# Simulating probability with a dice roll
# Let's simulate 10000 rolls of a fair 6-sided die
rolls = np.random.randint(1, 7, size=10000)
probability_of_3 = np.sum(rolls == 3) / len(rolls)

# Plotting the probability distribution of dice rolls
plt.figure(figsize=(8, 6))
sns.histplot(rolls, kde=False, bins=6, color='purple')
plt.title("Probability Distribution of Dice Rolls")
plt.xlabel("Dice Face")
plt.ylabel("Frequency")
plt.show()

# Display probability of rolling a 3
print(f"Probability of rolling a 3: {probability_of_3:.2f}")
```

#Difference Between Continuous and Discrete Distributions
Distributions describe how values of a random variable are spread or distributed.






Difference Between Continuous and Discrete Distributions (What is Distributions?)
Continuous Distributions: Variables can take any value within a continuous range (e.g., height, temperature). These are represented by smooth curves, and the probability density function (PDF) is used to describe them.

Discrete Distributions: Variables can only take distinct, countable values (e.g., the number of people in a room). These are represented by bar charts or probability mass functions (PMF).

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Continuous Distribution (Normal)
continuous_data = np.random.normal(0, 1, 1000)

# Discrete Distribution (Poisson)
discrete_data = np.random.poisson(5, 1000)

# Plotting both distributions
plt.figure(figsize=(12, 6))

# Continuous Distribution Plot
plt.subplot(1, 2, 1)
sns.histplot(continuous_data, kde=True, color='blue', bins=30)
plt.title("Continuous Distribution (Normal)")

# Discrete Distribution Plot
plt.subplot(1, 2, 2)
sns.histplot(discrete_data, kde=False, color='green', bins=30)
plt.title("Discrete Distribution (Poisson)")

plt.tight_layout()
plt.show()
```
