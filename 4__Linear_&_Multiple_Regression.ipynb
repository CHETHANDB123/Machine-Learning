###Label Encoder

```python
#Convert text to 0 & 1 [Binary form]
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
col=["driveway"	, "recroom"	, "fullbase"	, "gashw"	, "airco" , "prefarea"]
for f in col:
  df[f]=le.fit_transform(df[f])
df.head()
```

###Determing X & Y for SLR

```python
df.corr()
# Correlation quantifies the strength of the linear relationship b/w a pair of var
# Correlation of r=0.9 suggest a strong, positive association b/w 2 var
# Correlation of r=-0.2 suggest a weak, negative association.
```

##Defining what is X & Y for model
###Declaration & Inatialization

```python
x = df[["lotsize"]]
y = df[["price"]]
# Double brackets are used when we want to select 1 or more columns as a dataframe
# Single brackets are used when we want select single column
```

###Dividing into Train & Test Data

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2, random_state=2)
```

###Building simple Linear Regression Model

```python
from sklearn.linear_model import LinearRegression
slr = LinearRegression()
slr.fit(x_train,y_train)
```

```python
slr.coef_  # The slope is often called the reggresion coefficient
```

```python
slr.intercept_  # Constant value
```

###Evaluation

```python
ypredict = slr.predict(x_test)
ypredict
```

```python
#Accuracy Score
from sklearn.metrics import r2_score
r2_score(y_test,ypredict)
```

### Prediction for House price for given lotsize

```python
import numpy as np
x_new = np.array([5000]).reshape(1,-1)
slr.predict(x_new)
```

##Multiple Linear Regression

```python
#Multiple Linear Regression = for 1 'y' variable to take more than 1 'x' variable
#Single Linear Regression  = for 1 'y' variable take 1 'x' variable
```

###prediction of lotsize & bathrooms

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
#select the features & target var
x=df[["lotsize","bathrms"]]
y=df.price
#Split data into training & testing sets
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=2)
#create linear refression model & fit to the training Data
mlr=LinearRegression()
mlr.fit(x_train,y_train)

print(mlr.coef_)
print(mlr.intercept_)

ypredict=mlr.predict(x_test)
r2=r2_score(y_test,ypredict)
print("r2_score:" , r2)
#Deploy
xnew = np.array([[5000,2]])
prediction = mlr.predict(xnew)
print("prediction:",prediction)
```
