#Filling Null Values

### 1. Mannually Filling Null Values

```python
df1["SepalLengthCm"].fillna(7.2,inplace=True)
df1.head(15)
```

###2. Filling null Values with its Median

```python
med = df1["SepalWidthCm"].median()
df1["SepalWidthCm"]=df1["SepalWidthCm"].fillna(med)
df1.head(15)
```

###3.Fill Null Values with Mean

```python
me=df1["PetalLengthCm"].mean()
df1["PetalLengthCm"]=df1["PetalLengthCm"].fillna(me)
df1.head(10)
```
