```python
x=data[["Weight","Size"]]
y=data.Class
```

```python
from sklearn.tree import DecisionTreeClassifier
clf=DecisionTreeClassifier()
clf.fit(x,y)
```

```python
ypred=clf.predict(x)
```

```python
from sklearn.metrics import confusion_matrix
from sklearn.metrics import accuracy_score
print("confusion_matrix:",confusion_matrix(y,ypred))
print("accuracy_score:",accuracy_score(y,ypred))
```

###Plot Tree

```python
from sklearn.tree import plot_tree
plot_tree(clf,feature_names=["Weight","Size"],class_names=["orange","apple"])
```
