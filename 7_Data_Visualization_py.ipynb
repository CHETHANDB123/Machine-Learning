```python
import matplotlib.pyplot as plt
age=[5,10,15,20,25]
height_male=[50,90,120,180,185]
height_female=[45,80,115,160,165]
plt.title("Variation of height with age in males and females")
plt.xlabel("Age (in years)")
plt.ylabel("Height (in cm)")
plt.xlim(0,30)
plt.ylim(0,200)
plt.plot(age,height_male,linestyle="dashed",color="pink")
plt.plot(age,height_female,linestyle="dashdot",color='#008000')
plt.text(12,120,"Males")
plt.text(15,100,"Females")
plt.show()
```

##3D plots and Wordclouds

```python
from wordcloud import wordcloud,STOPWORDS, ImageColorGenerator
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
text="coke,pepsi,miranda,trip,read,write,coke,trip,data,visualization"
```

```python
sw=set(STOPWORDS)
sw
```

```python
sw.update(['data','visualization','said'])
```

```python
from wordcloud import WordCloud,STOPWORDS, ImageColorGenerator
cloud=WordCloud(background_color="White", stopwords=sw)
wordcloud=cloud.generate(text)
plt.imshow(wordcloud)
plt.axis("off")
plt.show()
```
