###STEP 2 : In Data Preprocessing , Convert Categorical Col to Numerical column

```python
#To Find Values In Column
df["Species"].value_counts()
```

###1. Manual Conersion

```python
df1=df.copy(deep=True)
mapping={"Iris-setosa":3, "Iris-versicolor":2 ,"Iris-virginica":1}
df1["Species"]=df1["Species"].map(mapping)
df1.head()
```

```python
df1.tail()
```

###1. label Encoder

```python
#It is convert each value in a column to number
df2=df.copy(deep=True)
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
df2["Species"]=le.fit_transform(df2["Species"])
df2.head()
```

```python
df2["Species"]=le.inverse_transform(df2["Species"])
df2.head()
```

###2. One Hot Encoding

```python
df3=df.copy(deep=True)
from sklearn.preprocessing import OneHotEncoder
ohe=OneHotEncoder()
x3=df3[["Species"]]    #[] use for multiple columns
ohe.fit_transform(x3).toarray()
```

###4. Dummy Variables

##Feature Scaling
##1. Min Max Scaler

```python
from sklearn.preprocessing import MinMaxScaler
df4=df.copy(deep=True)
x2=df4.iloc[:,0:5]  #iloc=integer location  # row,column
mms=MinMaxScaler(feature_range=(1,2))  # x_Scaled = (x-x_min)/(x_max-x_min)
x2_Scaled=mms.fit_transform(x2)
x2_Scaled
```

###2. Standard Scaler

```python
from sklearn.preprocessing import StandardScaler
df5=df.copy(deep=True)
x5=df5.iloc[:,0:5]
scaler=StandardScaler()
x5_Scaled=scaler.fit_transform(x5)
x5_Scaled
```

###3. Binarizer

```python
from sklearn.preprocessing import Binarizer
df6=df.copy(deep=True)
x6=df6.iloc[:,0:5]
bin=Binarizer(threshold=4)  # if the value is less than threshold value displays 0 else dispalys 1
x6_scaled=bin.fit_transform(x6)
x6_scaled
```
